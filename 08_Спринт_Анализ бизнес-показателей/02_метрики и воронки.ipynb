Сохраните логи событий из файла /datasets/funnel_prod_events.csv в переменной events. Сгруппируйте данные переменной events по столбцу event_name и посчитайте количество строк в каждой группе.

### 3 - 2

Сгруппируйте данные из переменной events по полю event_name и посчитайте количество уникальных пользователей в каждой группе, применяя метод agg(). Отсортируйте по убыванию.
Результат сохраните в переменной users_count и выведите на экран.

```python
users_count = events.groupby('event_name').agg({'uid':['nunique']}).sort_values(by= ('uid', 'nunique'), ascending=False)
users_count
```

```python
events = pd.read_csv('./events.csv')
events.head()
```

```python
users = events.pivot_table(index= 'uid', columns='event_name', values='event_time', aggfunc='min')
users.head()
```

### 4 - 3


Постройте воронку. Посчитайте:
сколько пользователей добавили товар в корзину после первого захода на сайт;
сколько пользователей начали оформлять заказ после добавления товара в корзину;
сколько из них оплатили заказ после оформления.
Сохраните результаты вычислений в переменных n_add_to_cart, n_checkout и n_payment.

```python
step_1 = ~users['pageview'].isna()
step_2 = step_1 & (users['add_to_cart'] > users['pageview'])
step_3 = step_2 & (users['checkout'] > users['add_to_cart'])
step_4 = step_3 & (users['payment'] > users['checkout'])
```

```python
n_pageview = users[step_1].shape[0]
n_add_to_cart = users[step_2].shape[0]
n_checkout = users[step_3].shape[0]
n_payment = users[step_4].shape[0]
```

```python
print('Посетителей:', n_pageview)
print('Добавили товар в корзину:', n_add_to_cart)
print('Начали оформлять заказ:', n_checkout)
print('Оплатили:', n_payment)
```
