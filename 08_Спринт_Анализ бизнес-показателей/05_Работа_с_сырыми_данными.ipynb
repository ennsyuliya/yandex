Срез ad_visits, который содержит посещения, совершённые из рекламных источников (ym:s:lastTrafficSource == 'ad')

Срез organic_visits, который содержит «органические» посещения, т. е. совершённые из поисковых систем (ym:s:lastTrafficSource == 'organic')

Название источника трафика хранится в колонке "ym:s:lastTrafficSource".

```python
ad_visits = logs_with_duration[logs_with_duration['ym:s:lastTrafficSource'] == 'ad']
# Сделайте срез посещений из поисковых систем
organic_visits = logs_with_duration[logs_with_duration['ym:s:lastTrafficSource'] == 'organic']

# Выведите 10 строчек среза посещений из рекламных источников
print(ad_visits.head(10))
# Выведите 10 строчек среза посещений из поисковых систем
print(organic_visits.head(10))
```

### 4

На одной визуализации отобразите две гистограммы:

Распределение длительности посещений, совершённых из рекламных источников (ym:s:lastTrafficSource == 'ad')

Распределение длительности посещений, совершённых из поисковых систем (ym:s:lastTrafficSource == 'organic')

Гистограммы должны охватывать данные в диапазоне от 0 до 2000 секунд.

Для отображения гистограмм с прозрачностью установите параметр alpha=0.5.

```python
# Гистограмма 1
ax = ad_visits.hist("ym:s:visitDuration", range=(0, 2000), alpha=0.5)

# Гистограмма 2
organic_visits.hist("ym:s:visitDuration", range=(0, 2000), alpha=0.5, ax=ax, grid=True)
```

### 5

Проверив гипотезу равенства двух средних, оцените, существует ли статистически значимое различие между средним временем посещений из рекламных источников и средним временем посещений из поисковых систем.

Критерий статистической значимости alpha задайте равным 0.05.

Если гипотеза отвергается, выведите сообщение "Отвергаем нулевую гипотезу", в противном случае — "Не удалось отвергнуть нулевую гипотезу".

```python
ad_visits_durations = ad_visits['ym:s:visitDuration']

organic_visits_durations = organic_visits['ym:s:visitDuration']
```

```python
alpha = 0.05

results = st.ttest_ind(
    ad_visits_durations,
    organic_visits_durations)

# Сделайте сравнение p-value с заданным alpha и вывод значения
if results.pvalue == alpha:
    print("Не удалось отвергнуть нулевую гипотезу")
else:
    print("Отвергаем нулевую гипотезу")
```
