3.4. В среднем пользователи разглядывают всплывающие сообщения на сайте 3 секунды. Дисперсия равна 0,25 секунд в квадрате. По правилу трёх сигм рассчитайте, сколько времени нужно показывать сообщение, чтобы его успело разглядеть 99% пользователей.

Результат на экран выведите так: Время показа сообщения ... .

```python
adv_mean = 3
adv_var = 0.25
adv_std = np.sqrt(adv_var) # рассчитайте стандартное отклонение

adv_time = (adv_mean+adv_std*3) # рассчитайте время показа пользователю всплывающего сообщения

print(f'Время показа сообщения', adv_time)
```

3.5. Ваша компания проводит викторину в торговом центре.

Из опыта проведения таких мероприятий известно, что в среднем правильно отвечают 3% всех участников. Стандартное отклонение 0,4%.

На викторину записалось 6 тысяч человек. Чтобы заготовить достаточное количество призов, нужно знать, сколько может быть победителей. По правилу трёх сигм найдите интервал 

количества победителей.

Результат выведите так: Интервал: ... - ...

```python
quiz_mean = 6000*0.03 #правильно ответивших в среднем
quiz_std = 6000*0.04   #стандартное отклонение от общего числа участников

# посчитайте нижнюю границу интервала
quiz_botton_line = (quiz_mean - quiz_std*3)
quiz_top_line = (quiz_mean + quiz_std*3)


print(f'Интервал', quiz_botton_line, '-', quiz_top_line)
```
